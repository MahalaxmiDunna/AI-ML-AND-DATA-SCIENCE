1. Z-Score Method

The Z-Score method measures how many standard deviations a data point is away from the mean. It is suitable for datasets that follow a normal (Gaussian) distribution. If a data point has a Z-score greater than +3 or less than −3, it is generally considered an outlier.

Formula:

Z=
X−μ
/
σ
	​


where:

X = Data value
μ = Mean of the dataset
σ = Standard deviation

Decision Rule:

If |Z| < 3, the value is considered normal.
If |Z| ≥ 3, the value is considered an outlier.

Advantages:

Simple and easy to implement.
Effective for normally distributed data.

Disadvantages:

Not suitable for skewed distributions.

```python
  #z-score for outliers
  for i in num_data.columns:
    mean=num_data[i].mean()
    std=num_data[i].std()
    z_score=(num_data[i]-mean)/std
   #Dropping the outliers
  outliers=num_data[abs(z_score)>3]
  print("percentage of outlier in ",i,"is",len(outliers)/len(num_data)*100)
  num_data=num_data[abs(z_score)<=3]
```
