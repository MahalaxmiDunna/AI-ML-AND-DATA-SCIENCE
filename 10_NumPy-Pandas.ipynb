Display Series Information

Given:

marks=pd.series([85,90,78,88,95])

Display:

- Number of elements

- data type

- shape

- index



```python
import pandas as pd
marks=pd.Series([85,90,78,88,95])
print(marks.dtype)
print(marks.size)
print(marks.shape)
print(marks.index)
```

```python
num=pd.Series([10,20,30,40,50,60,70])
print(num[:3])
print(num[-3:])
print(num[2:6])
```
