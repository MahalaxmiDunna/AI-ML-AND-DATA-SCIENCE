```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
```

```python
data=pd.read_csv('https://raw.githubusercontent.com/tkseneee/Dataset/master/k_circle_sales.csv')
```

```python
num_data=data.select_dtypes(include=np.number)
```

```python
num_data.max() - num_data.min()
```

```python
num_data_sd=(num_data['Item_Visibility']-num_data['Item_Visibility'].mean())
/num_data['Item_Visibility'].std()
```

```python
num_data_sd.describe()
```

```python
num_data_sd.head()
```

###standard Scalar
stdscr= X- mean / standard deviation





```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
num_data_sp=sc.fit_transform(num_data)
num_data_sp=pd.DataFrame(num_data_sp,columns=num_data.columns)
```

```python
num_data_sp.head()
```

- Standard Scalar (X- mean)/ Standard deviaton // Standardization / Z- Score Scaling
 - MinMax Scalar (X- Xmin)/ (Xmax - Xmin)
 - Robust Scalar (X-Xmedian)/(Q3-Q1)

```python
num_data_iWeight=(num_data['Item_Weight'] -num_data['Item_Weight'].min())
/(num_data['Item_Weight'].max()-num_data['Item_Weight'].min())
```

```python
num_data_iWeight.describe()
```

```python
from sklearn.preprocessing import MinMaxScaler
mx=MinMaxScaler()
num_data_minmx=mx.fit_transform(num_data)
num_data_minmx=pd.DataFrame(num_data_minmx,columns=num_data.columns)
```

### Robust scaler
 - Xscaled= (X- median)/ IQR

```python
iqr=num_data['Item_MRP'].quantile(0.75)-num_data['Item_MRP'].quantile(0.25)
num_data_rs=num_data['Item_MRP'].median()/iqr
```

```python
from sklearn.preprocessing import RobustScaler
rb = RobustScaler()
num_data_rb = rb.fit_transform(num_data)
num_data_rb = pd.DataFrame(num_data_rb, columns=num_data.columns, index=num_data.index)
```
